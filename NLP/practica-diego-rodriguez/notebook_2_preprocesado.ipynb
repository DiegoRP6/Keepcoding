## Preprocesado

Creamos la funcion `preprocess_text` con las tecnicas vistas en clase:
minusculas, quitar puntuacion y numeros, tokenizar, quitar stopwords y
lematizar.

## Librerias

```python
import re
import string

import pandas as pd

import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
```

## 1. Cargar el subset

```python
df = pd.read_csv('reviews_subset.csv')
df['title'] = df['title'].fillna('')
df['text'] = df['text'].fillna('')
df['review'] = df['title'] + ' ' + df['text']
df.head()
```

## 2. Funciones auxiliares para el preprocesado


```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def a_minusculas(text):
    return text.lower()

def quitar_numeros(text):
    return re.sub(r'\d+', ' ', text)

def quitar_puntuacion(text):
    for p in string.punctuation:
        text = text.replace(p, ' ')
    return text

def tokenizar(text):
    return text.split()

def quitar_stopwords(tokens):
    return [t for t in tokens if t not in stop_words and len(t) > 2]

def lematizar(tokens):
    return [lemmatizer.lemmatize(t) for t in tokens]
```

## 3. Funcion principal `preprocess_text`


```python
def preprocess_text(text):
    text = a_minusculas(text)
    text = quitar_numeros(text)
    text = quitar_puntuacion(text)
    tokens = tokenizar(text)
    tokens = quitar_stopwords(tokens)
    tokens = lematizar(tokens)
    return ' '.join(tokens)
```

## 4. Probar la funcion con un ejemplo

```python
ejemplo = df['review'].iloc[0]
print('ORIGINAL:')
print(ejemplo)
print()
print('PREPROCESADO:')
print(preprocess_text(ejemplo))
```

## 5. Aplicar a todo el subset

```python
df['clean_text'] = df['review'].apply(preprocess_text)


df = df[df['clean_text'].str.len() > 0].reset_index(drop=True)
print('Reviews tras preprocesado:', len(df))
df[['review', 'clean_text', 'sentiment']].head()
```

## 6. Guardar el dataset preprocesado

```python
df[['clean_text', 'sentiment']].to_csv('reviews_preprocessed.csv', index=False)
print('Guardado reviews_preprocessed.csv')
```
